# What is a File?

A file is a container used to store information permanently.

Examples:

- notes.txt
- students.json
- config.yaml
- report.pdf

Unlike variables, files remain available even after the program closes.

# Text Files

The simplest type of file is a Text File.

Examples:

- notes.txt
- diary.txt
- students.txt

Text files store plain text.

## Opening a File

Python uses the open() function.

Syntax:

open(filename, mode)

Common Modes:

| Mode | Meaning |
|--------|--------|
| r | Read |
| w | Write |
| a | Append |
| x | Create |

```python
file = open("sample.txt", "r")
```

# Reading Files

Suppose sample.txt contains:

Hello Students
Welcome to Python

Let's read it.

```python
file = open("sample.txt", "r")

content = file.read()

print(content)

file.close()
```

## Better Way

Python provides the with statement.

It automatically closes the file after use.

```python
with open("sample.txt", "r") as file:
    content = file.read()

print(content)
```

# Writing Files

We can create or overwrite files using write mode.

```python
with open("sample.txt", "w") as file:
    file.write("Hello Students")
```

## Append Mode

Append adds new content without removing old content.

```python
with open("sample.txt", "a") as file:
    file.write("\nWelcome to Python")
```

# JSON

JSON stands for:

JavaScript Object Notation

JSON is one of the most common formats used to store and exchange data.

Most APIs and web applications use JSON.

## Python Dictionary vs JSON

Python Dictionary:

{
    "name": "Rohan",
    "marks": 85
}

Looks very similar to JSON.

```python
import json

student = {
    "name": "Rohan",
    "marks": 85
}

json_string = json.dumps(student)

print(json_string)
```

## dumps()

Dictionary → JSON String

## loads()

JSON String → Dictionary

```python
import json

json_data = '{"name":"Rohan","marks":85}'

student = json.loads(json_data)

print(student)
```

# Mini Project : Student Database

Now let's save student information permanently using JSON.

```python
import json

students = [
    {"name": "Rohan", "marks": 85},
    {"name": "Mohan", "marks": 90},
    {"name": "Ajay", "marks": 75}
]

with open("students.json", "w") as file:
    json.dump(students, file, indent=4)

print("Student Data Saved")
```

```python
import json

with open("students.json", "r") as file:
    students = json.load(file)

print(students)
```
