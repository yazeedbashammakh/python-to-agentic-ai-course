## What is Exception Handling?

**Exception Handling** is a mechanism that allows a program to detect and respond to unexpected errors without terminating the entire application.

Instead of crashing, the program can decide what should happen when an unexpected situation occurs.

For example, an application might:

- Display a friendly error message.
- Ask the user to enter the input again.
- Skip the invalid operation.
- Continue running the rest of the program.

This helps us build applications that are more reliable, user-friendly, and professional.

## The Exception Handling Structure

Python handles exceptions using a combination of blocks that work together.

The most commonly used structure consists of:

- `try`
- `except`

These two blocks are designed to work together.

The `try` block identifies the code that might produce an exception.

The `except` block tells Python what to do if an exception occurs.

Without an `except` (or a `finally`) block, a `try` block is incomplete.

## The `try` Block

The `try` block contains the code that Python should attempt to execute.

Think of it as telling Python:

> "Try running this code."

If everything executes successfully, Python continues normally.

If an exception occurs, Python immediately stops executing the remaining statements inside the `try` block and transfers control to the matching `except` block.

## The `except` Block

The `except` block contains the code that should execute when an exception occurs inside the `try` block.

Instead of terminating the application, Python executes the code inside the `except` block.

Think of it as telling Python:

> "If something goes wrong while trying to execute the previous code, do this instead."

### Syntax

```python
try:
    # Code that might produce an exception

except:
    # Code to execute if an exception occurs
```

````python
try:
    age = int(input("Enter your age: "))

    print(f"You are {age} years old.")

except:
    print("Please enter a valid age.")
````

## How `try-except` Works

Whenever Python encounters a `try-except` structure, it follows these steps:

1. Execute the code inside the `try` block.
2. If no exception occurs, ignore the `except` block.
3. If an exception occurs, immediately stop executing the remaining code inside the `try` block.
4. Execute the `except` block.
5. Continue executing the rest of the program.

Understanding this execution flow is more important than memorizing the syntax.

````python
try:
    number = int(input("Enter a number: "))

    print(100 / number)

except:
    print("Something went wrong.")
````

## What is a `ValueError`?

A **ValueError** occurs when a function receives a value of the correct type but with an inappropriate value or format.

For example:

```python
int("hello")
```

Python expected a string representing a number, such as:

```
"25"
```

Instead, it received:

```
"hello"
```

Since `"hello"` cannot be converted into an integer, Python raises a **ValueError**.

Let's look at another example.

````python
try:
    number = int(input("Enter a number: "))

    print(100 / number)

except ZeroDivisionError:
    print("You cannot divide by zero.")
````

## What is a `ZeroDivisionError`?

A **ZeroDivisionError** occurs when a program attempts to divide a number by zero.

Since division by zero is mathematically undefined, Python raises this exception to prevent the program from producing an invalid result.

## Multiple `except` Blocks

Different operations can produce different exceptions.

## The `else` Block

The **`else` block** contains code that should execute **only if the `try` block completes successfully**.

If an exception occurs, Python skips the `else` block entirely.

Think of it as telling Python:

> "If everything inside the `try` block succeeds, then execute this code."

### Syntax

```python
try:
    # Risky code

except ExceptionType:
    # Handle the exception

else:
    # Execute only if no exception occurs
```

````python
try:
    amount = int(input("Enter withdrawal amount: "))

    print("Processing transaction...")

except ValueError:
    print("Please enter a valid amount.")

else:
    print("Transaction Successful.")
````

## How the `else` Block Works

Whenever Python encounters an `else` block, it follows these steps:

1. Execute the code inside the `try` block.
2. If an exception occurs, execute the matching `except` block.
3. Skip the `else` block.
4. Continue with the rest of the program.

If no exception occurs:

1. Execute the `try` block.
2. Skip the `except` block.
3. Execute the `else` block.
4. Continue with the rest of the program.

## When Should We Use `else`?

The `else` block is useful when some code should run **only after the `try` block succeeds**.

Some common examples include:

- Displaying a success message.
- Confirming that a transaction completed.
- Showing that data was processed successfully.
- Continuing to the next step of an application.

Using `else` keeps successful code separate from error-handling code, making programs easier to read and maintain.

## The `finally` Block

The **`finally` block** contains code that **always executes** before the program leaves the exception handling structure.

It doesn't matter whether:

- An exception occurs
- No exception occurs
- An exception is handled

The code inside the `finally` block will always run.

Think of it as telling Python:

> "No matter what happens, execute this code before moving on."

### Syntax

```python
try:
    # Risky code

except ExceptionType:
    # Handle the exception

else:
    # Execute if no exception occurs

finally:
    # Always execute
```

````python
try:
    amount = int(input("Enter withdrawal amount:"))

    print("Processing transaction...")

except ValueError:
    print("Please enter a valid amount.")

else:
    print("Transaction Successful.")

finally:
    print("Thank you for using our ATM.")
````

## The `raise` Statement

The **`raise`** statement allows us to create our own exceptions whenever one of our application's rules is violated.

## Why Do We Use `if` with `raise`?

You might have noticed that we placed the `raise` statement inside an `if` block.

This is because the exception should occur **only when a specific condition becomes true**.

In our example:

```python
if marks > 100:
```

checks whether the marks violate our application's rule.

If the condition is **True**, the `raise` statement creates an exception.

If the condition is **False**, the program continues normally.

The two statements have different responsibilities.

- The **`if` statement** checks whether a rule has been violated.
- The **`raise` statement** tells Python to treat that situation as an exception.

## Using `raise` with `try-except`

Just like Python's built-in exceptions, exceptions created using `raise` can also be handled using `try-except`.

````python
try:
    marks = int(input("Enter marks: "))

    if marks > 100:
        raise ValueError("Marks cannot exceed 100.")

except ValueError:
    print("Invalid Marks")
````

## Capturing the Exception

Whenever an exception occurs, Python creates an **Exception Object**.

This object stores information about the exception, such as:

- The type of exception.
- The error message.

If we want to access this information, we can store the exception object inside a variable.

The syntax is:

```python
except ExceptionType as variable_name:
```

Here,

```python
except ValueError as error:
```

`error` is simply a variable.

Python stores the exception object inside this variable so that we can use it inside the `except` block.

````python
try:
    marks = int(input("Enter marks: "))

    if marks > 100:
        raise ValueError("Marks cannot exceed 100.")

except ValueError as error:
    print(error)
````

Run the program.

### Test Case

```

```

Instead of displaying our own message, we're displaying the message stored inside the exception object.

## Understanding the `as` Keyword

The **`as`** keyword allows us to give the exception object a variable name.

Think of it exactly like assigning a value to a variable.

For example:

```python
name = "Rohan"
```

Here,

- `"Rohan"` is the value.
- `name` is the variable.

Similarly,

```python
except ValueError as error:
```

Here,

- The exception object is the value.
- `error` is the variable.

Once the exception object is stored inside the variable, we can use it like any other variable.

For example:

```python
print(error)
```

This prints the error message stored inside the exception object.

The variable name doesn't have to be `error`.

These examples are all valid:

```python
except ValueError as e:

except ValueError as exception:

except ValueError as my_error:
```

`error` is simply the most commonly used and descriptive name.
