# Object-Oriented Programming (OOP)

So far, we've learned how to:

Suppose a school asks us to build software to manage student records.

## Storing Student Information

One way to store a student's information is by using a dictionary.

````python
student = {
    "name": "Rohan",
    "roll": 101,
    "marks": 85
}

student
````

If we have multiple students, we can store them inside a list.

````python
students = [
    {
        "name": "Rohan",
        "roll": 101,
        "marks": 85
    },
    {
        "name": "Mohan",
        "roll": 102,
        "marks": 90
    },
    {
        "name": "Ajay",
        "roll": 103,
        "marks": 60
    },
    {
        "name": "Chandu",
        "roll": 104,
        "marks": 75
    },
    {
        "name": "Vijay",
        "roll": 105,
        "marks": 45
    }
]

students
````

So far, everything looks good.

Now let's start adding some features.

## Displaying Student Details

````python
def display(student):

    print(f"Name  : {student['name']}")
    print(f"Roll  : {student['roll']}")
    print(f"Marks : {student['marks']}")
````

````python
display(students[0])
````

Let's add another feature.

The school wants to update a student's marks.

````python
def update_marks(student, new_marks):

    student["marks"] = new_marks
````

````python
update_marks(students[0], 95)

display(students[0])
````

Next, let's check whether a student has passed.

````python
def is_pass(student):

    return student["marks"] >= 35
````

````python
print(is_pass(students[0]))
print(is_pass(students[4]))
````

Finally, let's calculate the student's grade.

````python
def calculate_grade(student):

    if student["marks"] >= 90:
        return "A"

    elif student["marks"] >= 75:
        return "B"

    elif student["marks"] >= 50:
        return "C"

    return "D"
````

````python
print(calculate_grade(students[0]))
print(calculate_grade(students[4]))
````

## Automatically Initializing Objects

Python provides a special method that runs automatically whenever a new object is created.

This special method is called the **Constructor**.

In Python, the constructor is written as:

```python
__init__()
```

Whenever we create an object using:

```python
Student(...)
```

Python automatically calls the constructor for us.

````python
class Student:

    def __init__(self, name, roll, marks):

        self.name = name
        self.roll = roll
        self.marks = marks
````

Now, instead of creating an empty object and adding attributes one by one, we can provide all the information while creating the object.

````python
student1 = Student("Rohan", 101, 85)
student2 = Student("Mohan", 102, 90)
student3 = Student("Ajay", 103, 60)
````

````python
print(student1.name)
print(student2.name)
print(student3.name)
````

Notice what changed.

Earlier, we wrote:

```python
student = Student()

student.name = ...
student.roll = ...
student.marks = ...
```

Now, we simply write:

```python
student = Student("Rohan", 101, 85)
```

The constructor automatically stores the values inside the object.

This makes object creation cleaner, shorter, and less error-prone.

## Understanding `self`

Take another look at the constructor.

```python
class Student:

    def __init__(self, name, roll, marks):

        self.name = name
        self.roll = roll
        self.marks = marks
```

One new keyword appears everywhere:

```python
self
```

What exactly is `self`?

Think about creating a student.

```python
student1 = Student("Rohan", 101, 85)
```

Python creates a new object for **Rohan**.

Inside the constructor, `self` refers to **that object**.

When we create another student,

```python
student2 = Student("Mohan", 102, 90)
```

`self` now refers to the new object created for **Mohan**.

In simple words:

> **`self` always refers to the object currently being created or used.**

Let's understand this line by line.

```python
self.name = name
```

There are two different `name`s here.

- `name` (right side) → The value we pass while creating the object.
- `self.name` (left side) → The attribute that belongs to the current object.

So when we write:

```python
student1 = Student("Rohan", 101, 85)
```

Python is essentially doing:

```python
student1.name = "Rohan"
student1.roll = 101
student1.marks = 85
```

automatically for us.

Likewise,

```python
student2 = Student("Mohan", 102, 90)
```

becomes:

```python
student2.name = "Mohan"
student2.roll = 102
student2.marks = 90
```

Each object stores its own information independently.

````python
student1 = Student("Rohan", 101, 85)
student2 = Student("Mohan", 102, 90)

print(student1.name)
print(student2.name)
````

Notice that both objects use the same constructor, but each object stores different values.

This is possible because `self` always refers to the current object.

## Why Don't We Write This?

You might wonder why we write:

```python
self.name = name
```

instead of simply writing:

```python
name = name
```

Let's see what happens.

````python
class Student:

    def __init__(self, name, roll, marks):

        name = name
        roll = roll
        marks = marks
````

````python
student1 = Student("Rohan", 101, 85)

print(student1.name)
````

Python raises an error:

```text
AttributeError: 'Student' object has no attribute 'name'
```

Why?

Because this line:

```python
name = name
```

only assigns the parameter back to itself.

It **does not create an attribute** inside the object.

To store information inside the object, we must write:

```python
self.name = name
```

The same applies to every other attribute.

Every object has its own attributes, and `self` helps Python know **which object's attributes** you're referring to.

## Functions Inside a Class

Let's move our `display()` function into the `Student` class.

````python
class Student:

    def __init__(self, name, roll, marks):

        self.name = name
        self.roll = roll
        self.marks = marks

    def display(self):

        print(f"Name  : {self.name}")
        print(f"Roll  : {self.roll}")
        print(f"Marks : {self.marks}")
````

Notice that `display()` now belongs to the `Student` class.

Functions that belong to a class are called **Methods**.

````python
student1 = Student("Rohan", 101, 85)

student1.display()
````

Look carefully at the way we called the method.

Earlier, we wrote:

```python

```

Now we write:

```python

```

This reads almost like English.

> "Student1, display your details."

Since the method already belongs to the object, we no longer need to pass the object manually.

Let's move another function into the class.

````python
class Student:

    def __init__(self, name, roll, marks):

        self.name = name
        self.roll = roll
        self.marks = marks

    def display(self):

        print(f"Name  : {self.name}")
        print(f"Roll  : {self.roll}")
        print(f"Marks : {self.marks}")

    def is_pass(self):

        return self.marks >= 35
````

````python
student1 = Student("Rohan", 101, 85)
student2 = Student("Vijay", 105, 25)

print(student1.is_pass())
print(student2.is_pass())
````

Notice something interesting.

Inside the method, we never wrote:

```python
student["marks"]
```

or

```python
student.marks
```

Instead, we wrote:

```python
self.marks
```

Why?

Because `self` already refers to the current object.

When we write:

```python
student1.is_pass()
```

Python automatically makes:

```python
self
```

refer to `student1`.

When we write:

```python
student2.is_pass()
```

`self` automatically refers to `student2`.

That's why the same method works for every object.

Let's move one more function into our class.

````python
class Student:

    def __init__(self, name, roll, marks):

        self.name = name
        self.roll = roll
        self.marks = marks

    def display(self):

        print(f"Name  : {self.name}")
        print(f"Roll  : {self.roll}")
        print(f"Marks : {self.marks}")

    def is_pass(self):

        return self.marks >= 35

    def update_marks(self, new_marks):

        self.marks = new_marks
````

````python
student1 = Student("Rohan", 101, 85)

student1.display()

student1.update_marks(95)

print()

student1.display()
````

Finally, let's move our grading function into the class.

````python
class Student:

    def __init__(self, name, roll, marks):

        self.name = name
        self.roll = roll
        self.marks = marks

    def display(self):

        print(f"Name  : {self.name}")
        print(f"Roll  : {self.roll}")
        print(f"Marks : {self.marks}")

    def is_pass(self):

        return self.marks >= 35

    def update_marks(self, new_marks):

        self.marks = new_marks

    def calculate_grade(self):

        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 50:
            return "C"

        return "D"
````

````python
student1 = Student("Rohan", 101, 85)

student1.display()

print("Pass :", student1.is_pass())
print("Grade:", student1.calculate_grade())

student1.update_marks(95)

print()

student1.display()

print("Pass :", student1.is_pass())
print("Grade:", student1.calculate_grade())
````

## Python's Solution

Python allows one class to reuse the properties and methods of another class.

This concept is called **Inheritance**.

Instead of writing the common code multiple times, we write it once and allow other classes to inherit it.

## Creating a Parent Class

Let's move the common information into a new class called `Person`.

````python
class Person:

    def __init__(self, name, id):

        self.name = name
        self.id = id

    def display(self):

        print(f"Name : {self.name}")
        print(f"ID   : {self.id}")
````

Now both Students and Teachers are also Persons.

Instead of writing everything again, they can inherit from the `Person` class.

````python
class Student(Person):
    pass


class Teacher(Person):
    pass
````

Notice the syntax.

```python
class Student(Person):
```

The name inside the brackets tells Python that `Student` inherits from `Person`.

Here:

- `Person` is the **Parent Class** (also called the Base Class).
- `Student` is the **Child Class** (also called the Derived Class).

Let's see whether a student can use the `display()` method from the parent class.

````python
student1 = Student("Rohan", 101)

student1.display()
````

It works!

Even though we never wrote a `display()` method inside the `Student` class, it was inherited from `Person`.

This is one of the biggest advantages of inheritance:

**Write once, reuse many times.**

## Adding Student-Specific Information

Students still need one extra attribute:

- Marks

How can we add it without rewriting everything from the `Person` class?

Python provides another useful function called `super()`.

````python
class Student(Person):

    def __init__(self, name, id, marks):

        super().__init__(name, id)

        self.marks = marks
````

### What is `super()`?

The `super()` function allows a child class to access the parent class.

Here,

```python
super().__init__(name, id)
```

calls the constructor of the `Person` class.

Instead of writing:

```python
self.name = name
self.id = id
```

again,

we simply reuse the parent's constructor.

After that, we only add what is unique to a student:

```python
self.marks = marks
```

````python
student1 = Student("Rohan", 101, 85)

student1.display()

print(student1.marks)
````

Notice that:

- `display()` comes from the `Person` class.
- `marks` belongs only to the `Student` class.

The student object now contains both inherited and its own attributes.

````python
class Teacher(Person):

    def __init__(self, name, id, salary):

        super().__init__(name, id)

        self.salary = salary
````

````python
teacher1 = Teacher("Mr. Sharma", 201, 50000)

teacher1.display()

print(f"Salary : ₹{teacher1.salary}")
````

## The Inheritance Hierarchy

```
           Person
          /      \
     Student   Teacher
```

Both `Student` and `Teacher` inherit the common features of a `Person`.

This avoids repeating the same code in multiple classes and makes our programs easier to maintain.

## Redefining an Inherited Method

Python allows a child class to replace a method inherited from its parent.

This is called **Method Overriding**.

Let's override the `display()` method in the `Student` class.

````python
class Student(Person):

    def __init__(self, name, id, marks):

        super().__init__(name, id)
        self.marks = marks

    def display(self):

        print(f"Name  : {self.name}")
        print(f"ID    : {self.id}")
        print(f"Marks : {self.marks}")
````

````python
student1 = Student("Rohan", 101, 85)

student1.display()
````

Notice something.

The `Person` class already had a `display()` method.

But Python executed the `Student` version instead.

The child class has **overridden** the parent's method.

This is known as **Method Overriding**.

Let's do the same for teachers.

````python
class Teacher(Person):

    def __init__(self, name, id, salary):

        super().__init__(name, id)
        self.salary = salary

    def display(self):

        print(f"Name   : {self.name}")
        print(f"ID     : {self.id}")
        print(f"Salary : ₹{self.salary}")
````

````python
teacher1 = Teacher("Mr. Sharma", 201, 50000)

teacher1.display()
````

Both classes inherited from `Person`, but each class now has its own version of `display()`.

```
Person
   │
display()

   ▲
   │
───────────────
│             │

Student    Teacher

display()  display()
```

Each child class can customize inherited methods whenever needed.

In simple words,

> **One interface, many different behaviors.**

## Can We Have Multiple Constructors?

In some programming languages like Java or C++, we can write multiple methods with the same name but different parameters.

This is called **Method Overloading**.

For example:

```text
calculate()

calculate(a)

calculate(a, b)
```

Does Python support this?

Let's try.

````python
class Calculator:

    def add(self, a):
        return a

    def add(self, a, b):
        return a + b
````

Python does **not** support true method overloading.

The second method replaces the first one.

If you run this code, only the last `add()` method exists.

Instead, Python achieves similar flexibility using:

- Default arguments
- `*args`
- `**kwargs`

which you've already learned.

````python
class Calculator:

    def add(self, *numbers):

        return sum(numbers)
````

````python
calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
````

Although this looks like method overloading, it isn't.

We are simply using `*args` to accept a variable number of arguments.

This is the common Pythonic way to achieve similar behavior.

## Controlling Access to Attributes

So far, every attribute we've created can be accessed directly.

```python
student.name
student.marks
```

Sometimes this is exactly what we want.

But sometimes we may want to protect certain data from being modified directly.

Python provides three levels of access.

````python
class Student:

    def __init__(self):

        self.name = "Rohan"
        self._roll = 101
        self.__marks = 85
````

### Public Attribute

```python
self.name
```

A public attribute can be accessed from anywhere.

```python
student.name
```

This is the default behavior in Python.

### Protected Attribute

```python
self._roll
```

A protected attribute starts with a single underscore (`_`).

It tells other developers:

> "This attribute is intended for internal use."

Python does not prevent access to it, but by convention, it should be treated as protected.

### Private Attribute

```python
self.__marks
```

A private attribute starts with two underscores (`__`).

Python makes it much harder to access directly from outside the class.

Private attributes are commonly used to hide sensitive data and encourage controlled access through methods.

````python
student = Student()

print(student.name)

print(student._roll)

print(student.__marks)
````

The last line raises an error because `__marks` is private.

We'll learn proper techniques for accessing private data when we build larger applications later in the course.

## Encapsulation

Throughout this notebook, we've been storing both **data** and the **methods that work on that data** inside the same class.

For example:

```python
class Student:

    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks

    def display(self):
        ...

    def calculate_grade(self):
        ...
```

Instead of keeping data and functions separate, they are grouped together inside one class.

This idea is called **Encapsulation**.

In simple words,

> **Encapsulation means keeping related data and methods together inside a single unit (a class).**

You've actually been using encapsulation throughout this notebook.

## Abstraction

Think about using a mobile phone.

You can:

- Make a call
- Send a message
- Take a photo

But you don't need to know how thousands of electronic components work inside the phone.

The complicated details are hidden from you.

This idea is called **Abstraction**.

In programming, abstraction means:

> **Showing only what is necessary while hiding the implementation details.**

As we build larger applications later in this course, you'll naturally start using abstraction without even realizing it.

## Packages

Until now, we've written all our code inside a single notebook.

As projects become larger, keeping everything in one file becomes difficult.

Python allows us to organize our code into multiple files.

For example:

```
school/
│
├── student.py
├── teacher.py
├── library.py
├── fees.py
└── main.py
```

Each file is called a **Module**.

A collection of related modules is called a **Package**.

Packages help us organize large applications into smaller, manageable parts.

We'll start creating our own modules and packages in the next few lessons.

| Pillar | Purpose |

| Polymorphism | One interface, different behaviors |
| Abstraction | Hide unnecessary implementation details |

In this notebook, we explored **Inheritance** in detail because it forms the foundation for writing reusable code.

Build a simple **Library Management System** using OOP.

Create a class named `Book`.

Each book should store:

- Title
- Author
- Available Copies

Your class should provide methods to:

- Display book details
- Borrow a book
- Return a book
- Check whether the book is available

After creating the class:

- Create at least three book objects.
- Borrow and return books.
- Display the updated details of every book.
